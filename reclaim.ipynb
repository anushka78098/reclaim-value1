```python
from pathlib import Path

root = Path('/content/codd')
xml_files = list(root.rglob('*.xml'))
image_suffixes = {'.jpg', '.jpeg', '.png'}
matched = sum(
    1 for xml in xml_files
    if any(xml.with_suffix(ext).exists() for ext in image_suffixes)
)

print("XML annotations found:", len(xml_files))
print("Annotations with a matching image:", matched)
```
